1. Crie uma lista com os valores [1, 4, 9, 10, 23] e execute as seguintes operações:

a) Obtenha o tamanho da lista

b) Obtenha o primeiro e depois o último elemento da lista

c) Utilizado slicing obtenha as sublistas [4, 9] e [10, 23]

d) Adicione o valor 90 ao fim da lista

e) Adicione o valor 5 entre o 10 e o 23

f) Remova o valor 9 da lista

g) Usando a keyword “in” verifique se os números 9 e 10 estão na lista (individualmente)

h) Ordene a lista por ordem crescente

i) Faça o reverso da lista


```
lista = [1, 4, 9, 10, 23]

# a) tamanho da lista
print(len(lista))

# b) primeiro e último elemento
print(lista[0])
print(lista[-1])

# c) sublistas [4, 9] e [10, 23]
print(lista[1:3])
print(lista[3:5])

# d) adicionar 90 ao fim
lista.append(90)
print(lista)

# e) adicionar 5 entre 10 e 23
lista.insert(4, 5)
print(lista)

# f) remover o valor 9
lista.remove(9)
print(lista)

# g) verificar se 9 e 10 estão na lista
print(9 in lista)
print(10 in lista)

# h) ordenar por ordem crescente
lista.sort()
print(lista)

# i) reverso da lista
lista.reverse()
print(lista)
```

2. Crie uma lista com os valores [0, 1, 2, 3, 4] e:

a) Usando “list comprehensions” crie uma nova lista com os quadrados dos números

b) À lista anterior calcule o somatório e a média

c) Faça novamente o mesmo, mas utilize a função range para criar uma lista 0..20

```
lista = [0, 1, 2, 3, 4]

# a) quadrados com list comprehension
quadrados = [x**2 for x in lista]
print(quadrados)

# b) soma e média
soma = sum(quadrados)
media = soma / len(quadrados)
print(soma)
print(media)

# c) usando range para criar 0..20
numeros = list(range(21))
quadrados = [x**2 for x in numeros]
print(quadrados)

soma = sum(quadrados)
media = soma / len(quadrados)
print(soma)
print(media)
```

3. Crie um dicionário com o nome e idade de alunos. Utilize o dicionário {“Pedro”: 10, “Ana”:
12, “Inês”: 11, “Paulo”: 9} e execute as seguintes operações:

a) Obtenha o número de alunos

b) Obtenha a idade do Pedro

c) Modifique a idade da Ana para 7

4. Faça um programa que receba uma string e conte a frequência de cada letra, colocando o
resultado num dicionário. Por exemplo, para a palavra “paralela” deverá imprimir as
seguintes frequências {'a': 3, 'l': 2, 'p': 1, 'r': 1, 'e': 1}. Sugestão: pesquise o método get(key
[, default]).

```
texto = input("Digite uma string: ")
freq = {}

for letra in texto:
    freq[letra] = freq.get(letra, 0) + 1

print(freq)
```

1. Implemente as seguintes funções:

a) add2(a, b) que retorna a soma de dois números

b) max(a, b) que retorna o maior entre dois números

c) is_divisible(a, b) que retorna verdadeiro ou falso consoante b divide a (divisão inteira)

d) avg(list) que retorna a média dos valores de uma lista

e) pow(a, b=2) que calcula ab , onde por defeito b=2 mas pode assumir outro valor

```
def add2(a, b):
    return a + b

def max(a, b):
    if a > b:
        return a
    else:
        return b

def is_divisible(a, b):
    return a % b == 0

def avg(lista):
    return sum(lista) / len(lista)

def pow(a, b=2):
    return a ** b
```

2. Implemente as seguintes funções de forma recursiva:

a) Factorial, sabendo que f(0)=1 e f(n) = n*f(n-1)

b) Somatório dos primeiros n números, sabendo que s(0]) = 0 e s(n) =n+s(n-1)

c) Expoente, sabendo que pow(n, 0) = 1 e pow(n, x) = n * pow(n, x-1)

```
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

def somatorio(n):
    if n == 0:
        return 0
    else:
        return n + somatorio(n - 1)

def potencia(n, x):
    if x == 0:
        return 1
    else:
        return n * potencia(n, x - 1)
```

1. Use o módulo “math” (https://docs.python.org/3/library/math.html) para obter:

a) Factorial de 10.

b) Máximo divisor comum entre 6 e 3.

c) Número de combinações de 20, 2 a 2.

d) Número de permutações de 10.

e) O log2 de 25 e o log10 de 60.

f) O coseno de math.pi.

g) A distância entre os pontos (0,0) e (10,10).


```
import math

# a) Factorial de 10
print(math.factorial(10))

# b) Máximo divisor comum entre 6 e 3
print(math.gcd(6, 3))

# c) Número de combinações de 20, 2 a 2
print(math.comb(20, 2))

# d) Número de permutações de 10
print(math.perm(10))

# e) log2 de 25 e log10 de 60
print(math.log2(25))
print(math.log10(60))

# f) coseno de math.pi
print(math.cos(math.pi))

# g) Distância entre (0,0) e (10,10)
print(math.dist((0, 0), (10, 10)))
```

4. Faça um programa que gere um número aleatório entre 1 e 100 e peça números ao utilizador até acertar. Deverá dar sugestões do género “o número é maior”, “o número é menor”, etc.

```
import random

numero_secreto = random.randint(1, 100)
tentativa = 0

while tentativa != numero_secreto:
    tentativa = int(input("Adivinha o número (1 a 100): "))

    if tentativa < numero_secreto:
        print("O número é maior.")
    elif tentativa > numero_secreto:
        print("O número é menor.")
    else:
        print("Acertaste!")
```
