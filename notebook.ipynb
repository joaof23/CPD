2. Defina as variáveis “pi” e o raio “r” e calcule a área de um círculo (pi*r^2).

```python
import math

def circulo(r):
    return 2 * math.pi * r

r = float(input("Digite o raio: "))
print(circulo(r))
```

3. Defina uma string “s” com o conteúdo “abc” e execute as seguintes operações:

a) Obtenha o tamanho da string

b) Usando slice ([x:y]), obtenha os primeiros 2 caracteres da string

c) Usando slice obtenha os dois últimos caracteres

d) Usando indexação e concatenação, transforme a string em “aaabbbccc”

```python
s="abc"
print(len(s))
print(s[0:2]) #0 inclusivo 2 exclusivo
print(s[1:3]) 
resultado = s[0]*3 + s[1]*3 + s[2]*3
print(resultado)
```

4. Crie uma string com o conteúdo “aaabbbccc” e execute as seguintes operações:

a) Capitalize a string (“Aaabbbccc”)

b) Encontre o índice da primeira ocorrência de “b” e a primeira ocorrência de “ccc”

c) Altere todas as ocorrências de “a” para “X”, e depois apenas a primeira ocorrência

d) Transforme a string em “AAAbbbCCC” e depois em “AAABBBCCC”

```python
ns="aaabbbccc"
print(ns.capitalize())
print(ns.find("b"))
print(ns.find("ccc"))
print(ns.replace("a", "X"))
print(ns.replace("a", "X", 1))
print(ns[:3].upper() + ns[3:6] + ns[6:].upper())
```
